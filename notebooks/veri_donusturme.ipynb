```python
import os
import fitz  # pymupdf

pdf_path = "../data/raw/turkce.pdf"
output_txt = "../data/processed/turkce.txt"

def extract_two_column_text(page):
    blocks = page.get_text("blocks")  # (x0,y0,x1,y1,text,block_no,block_type)
    mid = page.rect.width / 2

    left, right = [], []
    for b in blocks:
        x0, y0, x1, y1, txt, *_ = b
        if not txt or not txt.strip():
            continue
        (left if x0 < mid else right).append((x0, y0, txt.strip()))

    left = sorted(left, key=lambda t: (t[1], t[0]))
    right = sorted(right, key=lambda t: (t[1], t[0]))

    return "\n".join([t[2] for t in left + right])

doc = fitz.open(pdf_path)

text = ""
for i in range(doc.page_count):
    page = doc.load_page(i)
    text += f"\n\n===== PAGE {i+1} =====\n\n"
    text += extract_two_column_text(page)

os.makedirs("../data/processed", exist_ok=True)
with open(output_txt, "w", encoding="utf-8") as f:
    f.write(text)

print("PDF metne dönüştürüldü ve kaydedildi:", output_txt)
```
