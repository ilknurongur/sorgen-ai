JSONL: JSON Lines <br>
Her satırda bir adet JSON obje bulunur. <br>
Örneğin dil modeli eğitimi için şöyle görünür: <br>
{"prompt": "Soru: Teknolojinin hayatımıza kattığı kolaylıklar...", "response": "Cevap: A"} <br>
{"prompt": "Soru: Bir toplumun sanat anlayışı...", "response": "Cevap: A"} <br>
Bu format özellikle T5, LLaMA, GPT-neoX, Falcon, Mistral, Gemma, LoRA fine-tuning gibi modeller için idealdir.
### ------------------------------------------------------------------------------------------------------------
Eğer büyük bir veriyi klasik .json ile açarsan: tüm veriyi belleğe yüklemeye çalışır, RAM taşar. Ama .jsonl dosyası ile okunursa satır satır, yani akış halinde (streaming) okunur. Bu, özellikle model eğitimi(fine-tuning) sırasında GPU'ya veri akışı için mükemmeldir.

```python
import json
from pathlib import Path

input_path  = Path("../data/processed/turkce_questions.json")
output_path = Path("../data/processed/turkce_questions.jsonl")

INCLUDE_METADATA = True
SKIP_IF_NO_ANSWER = False
OPTIONS_ORDER = ["A","B","C","D","E"]

def one_line(s: str) -> str:
    if s is None:
        return ""
    s = s.replace("\r", " ").replace("\n", " ")
    s = " ".join(s.split())
    return s.strip()

with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

written = 0
skipped_answer = 0
skipped_year = 0

with open(output_path, "w", encoding="utf-8") as f:
    for item in data:
        yil = item.get("yil", None)

        # ✅ YIL BOŞSA YAZMA
        if not yil:
            skipped_year += 1
            continue

        soru = one_line(item.get("soru", ""))
        secenekler = item.get("secenekler", {}) or {}
        cevap = item.get("cevap", None)

        if SKIP_IF_NO_ANSWER and not cevap:
            skipped_answer += 1
            continue

        opts_txt_parts = []
        for k in OPTIONS_ORDER:
            v = secenekler.get(k)
            if v:
                opts_txt_parts.append(f"{k}) {one_line(v)}")
        options_text = "\n".join(opts_txt_parts)

        soru_no = item.get("soru_no")

        meta = ""
        if INCLUDE_METADATA:
            meta_bits = [f"Yıl: {yil}"]
            if soru_no:
                meta_bits.append(f"Soru No: {soru_no}")
            meta = " | ".join(meta_bits)

        prompt_lines = []
        if meta:
            prompt_lines.append(f"[{meta}]")
        prompt_lines.append(f"Soru: {soru}")
        prompt_lines.append("Seçenekler:")
        prompt_lines.append(options_text)
        prompt_lines.append("Doğru seçenek hangisidir? ")
        prompt = "\n".join(prompt_lines).strip()

        response = (cevap or "").strip()

        out = {
            "prompt": prompt,
            "response": response,
            "id": item.get("id"),
            "yil": yil,
            "soru_no": soru_no
        }

        f.write(json.dumps(out, ensure_ascii=False) + "\n")
        written += 1

print(f"✅ JSONL oluşturuldu: {output_path}")
print(f"Yazılan kayıt: {written}")
print(f"Atlanan (yıl boş): {skipped_year}")
print(f"Atlanan (cevap boş): {skipped_answer}")
```
